```python
# Problem 1: Villager Class
# A class constructor is a special method or function that is used to create and initialize a new object from a class. Define the class constructor __init__() for a new class Villager that represents characters in the game Animal Crossing. The constructor accepts three required arguments: strings name, species, and catchphrase. The constructor defines four properties for a Villager:

```

```python
class Villager:

    def __init__(self, name, species, personality, catchphrase, neighbor=None):
        self.name = name
        self.species = species
        self.personality = personality
        self.catchphrase = catchphrase
        self.furniture = []
        self.neighbor = neighbor
        self.validFurniture = set(["acoustic guitar", "ironwood kitchenette", "rattan armchair", "kotatsu", "cacao tree"])

    # Function that adds a piece of furniture to the furniture list
    def add_item(self, item_name):
        if item_name in self.validFurniture:
            self.furniture.append(item_name)

# Returns list of names of villagers with matching personality type    
def of_personality_type(townies, personality_type):
    villagerRet = []
    for villager in townies: # O(n) time and space complexity
        if villager.personality == personality_type:
            villagerRet.append(villager.name)
    return villagerRet

# Returns True if message can be passed from start_villager to target_villager through neighbors
def message_received(start_villager, target_villager):
    curr = start_villager
    while curr:
        if curr == target_villager:
            return True
        curr = curr.neighbor
    return False

class Node:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next

# For testing
def print_linked_list(head):
    current = head
    while current:
        print(current.value, end=" -> " if current.next else "\n")
        current = current.next

# Prints name of head in message and removes first node in the list, if list empty return None and print message
def catch_fish(head):
    if head == None:
        print("Aw! Better luck next time!")
        return
    print(f"Congratulations you caught {head.value}")
    head = head.next
    return head

```

```python
class Node:
    def __init__(self, fish_name, next=None):
        self.fish_name = fish_name
        self.next = next

# For testing
def print_linked_list(head):
    current = head
    while current:
        print(current.fish_name, end=" -> " if current.next else "\n")
        current = current.next

def fish_chances(head, fish_name):
    count = 0
    found = 0
    curr = head
    while curr:
        count += 1
        if curr.fish_name == fish_name:
            found += 1
        curr = curr.next
    if(count != 0):
        return round(found/count,2)
    
    return 0
        

    

```

```python
fish_list = Node("Carp", Node("Dace", Node("Cherry Salmon")))
print(fish_chances(fish_list, "Dace"))
print(fish_chances(fish_list, "Rainbow Trout"))
```
