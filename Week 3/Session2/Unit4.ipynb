```python
def find_most_frequent_keywords(scenes):
    frequency_dict = {}
    for l in scenes.values():
        for val in l:
            if val in frequency_dict:
                frequency_dict[val] += 1
            else:
                frequency_dict[val] = 1
    maxVal = max(frequency_dict.values())
    ret = []
    for val in frequency_dict:
        if frequency_dict[val] == maxVal:
            ret.append(val)
    # Time complexity O(n*m) Space Complexity: O(n)
    return ret

scenes = {
    "Scene 1": ["action", "hero", "battle"],
    "Scene 2": ["hero", "action", "quest"],
    "Scene 3": ["battle", "strategy", "hero"],
    "Scene 4": ["action", "strategy"]
}
print(find_most_frequent_keywords(scenes))

scenes = {
    "Scene A": ["love", "drama"],
    "Scene B": ["drama", "love"],
    "Scene C": ["comedy", "love"],
    "Scene D": ["comedy", "drama"]
}
print(find_most_frequent_keywords(scenes)) 

# ['action', 'hero']
# ['love', 'drama']
```
